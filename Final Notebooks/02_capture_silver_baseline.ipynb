```python
# CAPTURE PREVIOUS SILVER STATE

from pyspark.sql import functions as F

silver_table = "final_assesment.dbt_dev.silver_orders"

if spark.catalog.tableExists(silver_table):

    previous_silver = spark.table(silver_table)

    previous_order_ids = [
        row["order_id"]
        for row in (
            previous_silver
            .select("order_id")
            .filter(F.col("order_id").isNotNull())
            .distinct()
            .collect()
        )
    ]

else:
    previous_order_ids = []

previous_silver_count = len(previous_order_ids)

# Pass the baseline to the downstream validation task
dbutils.jobs.taskValues.set(
    key="previous_order_ids",
    value=previous_order_ids
)

dbutils.jobs.taskValues.set(
    key="previous_silver_count",
    value=previous_silver_count
)

print(f"Previous Silver order count captured: {previous_silver_count}")
```
